# 01 — Project scope and protocol choices

## What this notebook does

This project compares three staking models:

1. **Native staking** — staking ETH directly  
2. **Liquid staking** — staking through a protocol and receiving a liquid token  
3. **Restaking** — using already staked assets in an additional reward layer  

The goal of this notebook is **only** to define the scope clearly:
- what we compare,
- which protocols represent each category,
- and why these choices make sense.

We keep this notebook simple on purpose.  
It is a **project definition notebook**, not a data collection notebook.

## Why these categories?

We want to study how staking evolved from a simple locked position into more flexible and more layered forms.

The main comparison dimensions of the project are:

- **Yield** — what return is offered?
- **Adoption / size** — how much capital is in the protocol?
- **Liquidity** — is the position locked or can it be reused/traded?
- **Complexity** — how many protocol layers are involved?

We will not try to model every detail of DeFi.  
Instead, we compare a small number of representative examples.
